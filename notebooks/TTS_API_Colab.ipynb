## Mở API với ngrok
Để truy cập API từ bên ngoài Colab, chúng ta sẽ sử dụng ngrok. Điền authtoken của bạn từ https://dashboard.ngrok.com/get-started/your-authtoken vào bên dưới.

```python
from pyngrok import ngrok

# Thay 'YOUR_AUTHTOKEN_HERE' bằng token thật của bạn
ngrok.set_auth_token("YOUR_AUTHTOKEN_HERE")

# Mở kết nối ngrok ở cổng 8000 (cổng mặc định của FastAPI)
public_url = ngrok.connect(8000)
print("===========================================================")
print(f"API của bạn đang chạy tại: {public_url}")
print(f"Tài liệu API (Swagger UI): {public_url}/docs")
print("===========================================================")
```
